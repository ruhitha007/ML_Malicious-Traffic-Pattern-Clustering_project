```python
import os

folder_path = '/content/CICIDS2017/MachineLearningCVE'  #Inspecting how many rows and columns are there in each csv file

for file in os.listdir(folder_path):
    if file.endswith('.csv'):
        df = pd.read_csv(os.path.join(folder_path, file))
        print(file, "->", df.shape)
```

```python
# Check whether all CSV files have the same columns

columns = {}

for file in os.listdir(folder_path):
    if file.endswith('.csv'):
        df = pd.read_csv(os.path.join(folder_path, file), nrows=5)
        columns[file] = list(df.columns)

first_file = list(columns.keys())[0]

for file, cols in columns.items():
    print(file, "->", cols == columns[first_file])
```

```python
# What types of network traffic are actually present in each raw CSV?
for file in os.listdir(folder_path):
    if file.endswith('.csv'):
        df = pd.read_csv(os.path.join(folder_path, file))
        label_column = df.columns[-1]

        print("\n", file)
        print(df[label_column].value_counts())
```

```python
# Check missing values in all CSV files

missing_summary = {}

for file in os.listdir(folder_path):
    if file.endswith('.csv'):
        df = pd.read_csv(os.path.join(folder_path, file))
        missing_summary[file] = df.isnull().sum().sum()

for file, missing in missing_summary.items():
    print(file, "->", missing, "missing values")
```

```python
# Check infinite values in all CSV files

infinite_summary = {}

for file in os.listdir(folder_path):
    if file.endswith('.csv'):
        df = pd.read_csv(os.path.join(folder_path, file))

        numeric_df = df.select_dtypes(include=np.number)
        infinite_count = np.isinf(numeric_df).sum().sum()

        infinite_summary[file] = infinite_count

for file, count in infinite_summary.items():
    print(file, "->", count, "infinite values")
```

```python
# Check duplicate rows in each CSV file

duplicate_summary = {}

for file in os.listdir(folder_path):
    if file.endswith('.csv'):
        df = pd.read_csv(os.path.join(folder_path, file))

        duplicate_count = df.duplicated().sum()
        duplicate_summary[file] = duplicate_count

for file, count in duplicate_summary.items():
    print(file, "->", count, "duplicate rows")
```

```python
# Create a sample from each CSV for basic statistical understanding

sample_list = []

for file in os.listdir(folder_path):
    if file.endswith('.csv'):
        df = pd.read_csv(os.path.join(folder_path, file))
        sample = df.sample(n=min(10000, len(df)), random_state=42)
        sample_list.append(sample)

sample_df = pd.concat(sample_list, ignore_index=True)

sample_df.describe().T
```

```python
# Check for features with zero variance in the sample

zero_variance = sample_df.select_dtypes(include=np.number).nunique()

zero_variance[zero_variance == 1]
```

```python
# Check the exact names of the suspected constant features

df_check = pd.read_csv(
    os.path.join(folder_path, 'Monday-WorkingHours.pcap_ISCX.csv'),
    nrows=1
)

for col in df_check.columns:
    if 'Bwd' in col or 'Fwd Avg' in col:
        print(repr(col))
```

```python
constant_features = [
    'Bwd PSH Flags',
    'Bwd URG Flags',
    'Fwd Avg Bytes/Bulk',
    'Fwd Avg Packets/Bulk',
    'Fwd Avg Bulk Rate',
    'Bwd Avg Bytes/Bulk',
    'Bwd Avg Packets/Bulk',
    'Bwd Avg Bulk Rate'
]

for file in os.listdir(folder_path):
    if file.endswith('.csv'):

        # Read only the column names
        headers = pd.read_csv(
            os.path.join(folder_path, file),
            nrows=0
        ).columns

        # Map stripped names to their original raw names
        column_map = {col.strip(): col for col in headers}

        # Get the original names from the raw CSV
        actual_columns = [
            column_map[col] for col in constant_features
            if col in column_map
        ]

        df = pd.read_csv(
            os.path.join(folder_path, file),
            usecols=actual_columns
        )

        print("\n", file)
        print(df.nunique())
```
